### Descripción de los datos

- **Customer**: identificador del cliente.

- **ST**: estado en el que viven los clientes.

- **Gender**: género del cliente.

- **Education**: nivel educativo del cliente.

- **Customer Lifetime Value**: el valor de vida del cliente (*Customer Lifetime Value*, CLV) es el ingreso total que la empresa espera obtener de un cliente durante toda su relación con él. En otras palabras, es el valor previsto o calculado de un cliente durante todo el tiempo que mantenga una póliza con la compañía de seguros. Es una estimación del beneficio neto que la aseguradora espera generar a lo largo de esa relación. El valor de vida del cliente tiene en cuenta factores como la duración de la póliza, los pagos de las primas, el historial de reclamaciones, la probabilidad de renovación y la posibilidad de que el cliente contrate servicios o productos adicionales. Ayuda a las aseguradoras a evaluar la rentabilidad a largo plazo y el valor de conservar a un cliente.

- **Income**: ingresos del cliente.

- **Monthly Premium Auto**: cantidad que el cliente paga mensualmente por la prima de su seguro de automóvil. Representa el coste recurrente que la persona asegurada debe pagar para mantener la póliza y contar con cobertura frente a posibles daños, accidentes u otros sucesos relacionados con su vehículo que estén cubiertos por el seguro.

- **Number of Open Complaints**: número de reclamaciones que ha abierto el cliente.

- **Policy Type**: existen tres tipos de pólizas de seguro de automóvil: *Corporate Auto*, *Personal Auto* y *Special Auto*.

- **Vehicle Class**: categoría del vehículo del cliente: coche de dos puertas, coche de cuatro puertas, SUV, SUV de lujo, coche deportivo y coche de lujo.

- **Total Claim Amount**: suma de todas las reclamaciones presentadas por el cliente. Representa el valor monetario total de las reclamaciones aprobadas por incidentes como accidentes, robos, actos vandálicos u otros sucesos cubiertos por la póliza.

#Calcula estadísticas descriptivas, como la media, la mediana, la moda, la desviación estándar y los cuartiles, para comprender la tendencia central y la distribución de los datos de las columnas numéricas. Expón también tus conclusiones a partir de estas estadísticas.

```python
data.head()
```

```python
#Calculo Media Income, Mediana Income, Moda Income, Desviacion Estandar Income, Q1, Q2 y Q3
print(data['Income'].mean())
print(data['Income'].median())
print(data['Income'].mode())
print(data['Income'].std())
q1 = data['Income'].quantile(0.25)
q2 = data['Income'].quantile(0.50)
q3 = data['Income'].quantile(0.75)
print(q1, q2, q3)
```

El ingreso medio (Income) de los clientes es aproximadamente 39.295,70 unidades monetarias.
La mitad de los valores de ingreso está por debajo o en 36.234, y la otra mitad por encima o en ese valor.
El ingreso que aparece con mayor frecuencia es 0.
La desviación estándar es de 30.469,43, lo que refleja una dispersión considerable de los ingresos.
El primer cuartil (Q1) es 14.072 y el tercer cuartil (Q3) es 64.631, por lo que el 50 % central de los ingresos se encuentra aproximadamente entre esos dos valores.
Conclusión general: Los ingresos presentan una dispersión considerable. La media es superior a la mediana, lo que podría indicar que algunos ingresos elevados aumentan el promedio. Además, el valor cero es el que aparece con mayor frecuencia, por lo que convendría revisar esos registros para comprender qué representan.



```python
#Calculo Media Total Claim Amount, Mediana Total Claim Amount, Moda Total Claim Amount, Desviacion Estandar Total Claim Amount, Q1, Q2 y Q3
print(data['Total Claim Amount'].mean())
print(data['Total Claim Amount'].median())
print(data['Total Claim Amount'].mode())
print(data['Total Claim Amount'].std())
q1 = data['Total Claim Amount'].quantile(0.25)
q2 = data['Total Claim Amount'].quantile(0.50)
q3 = data['Total Claim Amount'].quantile(0.75)
print(q1, q2, q3)
```
